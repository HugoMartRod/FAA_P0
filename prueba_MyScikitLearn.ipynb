```python
import pandas as pd
from MyScikitLearn.preprocesamiento import Dataset
from MyScikitLearn.particionado import ValidacionSimple, ValidacionCruzada

# Cargamos ambos datasets (asumiendo que están en la misma carpeta que el notebook)
dataset_ttt = Dataset('tic-tac-toe.csv')
dataset_heart = Dataset('heart.csv')
```

```python
print("================ TIC-TAC-TOE ================")
# 3.1 Datos originales
print("\n--- 3.1 Primeras 5 líneas (Original) ---")
display(dataset_ttt.datos_originales.head())

# 3.2 Datos transformados
print("\n--- 3.2 Primeras 5 líneas (Transformado) ---")
display(dataset_ttt.datos_numericos.head())

# 3.3 Diccionarios
print("\n--- 3.3 Diccionario de transformación ---")
# Mostramos un par de atributos para no saturar la pantalla
for col in dataset_ttt.datos_originales.columns[:2]:
    print(f"Columna '{col}': {dataset_ttt.diccionarios[col]}")
```

```python
print("================ HEART DISEASE ================")
# 3.1 Datos originales
print("\n--- 3.1 Primeras 5 líneas (Original) ---")
display(dataset_heart.datos_originales.head())

# 3.2 Datos transformados
print("\n--- 3.2 Primeras 5 líneas (Transformado) ---")
display(dataset_heart.datos_numericos.head())

# 3.3 Diccionarios (Veremos que hay columnas numéricas con diccionario vacío)
print("\n--- 3.3 Diccionario de transformación ---")
for col in dataset_heart.datos_originales.columns[:3]:
    print(f"Columna '{col}': {dataset_heart.diccionarios[col]}")
```

```python
print("================ PARTICIONAMIENTO (Tic-Tac-Toe) ================")
num_filas = dataset_ttt.datos_numericos.shape[0]

# 3.5 Validación Simple
val_simple = ValidacionSimple(proporcion_test=0.3, numero_ejecuciones=1)
val_simple.creaParticiones(num_filas)

particion_simple = val_simple.particiones[0]
print("\n--- 3.5 Validación Simple (Hold-out) ---")
print("Índices Train (primeros 15):", particion_simple["train"][:15])
print("Índices Test (primeros 15):", particion_simple["test"][:15])
print(f"Total Train: {len(particion_simple['train'])} | Total Test: {len(particion_simple['test'])}")
```

```python
# 3.4 Validación Cruzada
val_cruzada = ValidacionCruzada(numero_particiones=5)
val_cruzada.creaParticiones(num_filas)

particion_cruzada = val_cruzada.particiones[0]
print("\n--- 3.4 Validación Cruzada (K-Fold, partición 1) ---")
print("Índices Train (primeros 15):", particion_cruzada["train"][:15])
print("Índices Test (primeros 15):", particion_cruzada["test"][:15])
print(f"Total Train: {len(particion_cruzada['train'])} | Total Test: {len(particion_cruzada['test'])}")
```
